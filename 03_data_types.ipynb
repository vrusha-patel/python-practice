Some of the Data Types 

```python

name = "Sunita"
age = 20
height = 5.4
is_student = True

print(name)
print(age)
print(height)
print(is_student)

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

List Data Type

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]

print("Fruit List:")
print(fruits)

print("First Fruit:", fruits[0])

print(type(fruits))
```

Tuple Data Type

```python
colors = ("Red", "Green", "Blue")

print(colors)

print(type(colors))
```

Set Data Type

```python
numbers = {10, 20, 30, 20, 40}

print(numbers)

print(type(numbers))
```

Dictionary Data Type

```python
student = {
    "Name": "Sunita",
    "Age": 20,
    "Course": "Data Science"
}

print(student)

print(student["Name"])

print(type(student))
```

Checking user input data

```python
value = input("Enter something: ")

print("You entered:", value)
print("Data Type:", type(value))
```
