Implicit Type Casting (Automatic Conversion)
Definition

Python automatically converts a smaller compatible data type into a larger compatible data type during an operation.

```python
#Task 1: Add an Integer and a Float
a = 10
b = 5.5

print("Type of a:", type(a))
print("Type of b:", type(b))

result = a + b

print("Addition:", result)
print("Type of result:", type(result))
```

```python
#Task 2: Multiply an Integer with a Float
a = 8
b = 2.5

print("Type of a:", type(a))
print("Type of b:", type(b))

result = a * b

print("Multiplication:", result)
print("Type of result:", type(result))
```

```python
#Task 3: Divide Two Integers
a = 15
b = 2

print("Type of a:", type(a))
print("Type of b:", type(b))

result = a / b

print("Division:", result)
print("Type of result:", type(result))
```

```python
#Task 4: Check Type Before and After Every Operation
x = 20
y = 4.0

print("Before Operation")
print("Type of x:", type(x))
print("Type of y:", type(y))

z = x + y

print("\nAfter Operation")
print("Value of z:", z)
print("Type of z:", type(z))
```

Explicit Type Casting (Manual Conversion)
Definition

The programmer manually converts one data type into another using built-in functions.
Functions
int()
float()
str()
bool()
list()
tuple()
set()
dict()

```python
#int() Example
x = 5.9
y = int(x)

print(y)
print(type(y))
```

```python
#float() Example
x = 10
y = float(x)

print(y)
print(type(y))
```

```python
#str() Example
x = 100
y = str(x)

print(y)
print(type(y))
```

```python
#bool() Example
print(bool(1))
print(bool(0))
print(bool(""))
print(bool("Python"))
```

```python
#list() Example
x = (10, 20, 30)

y = list(x)

print(y)
print(type(y))
```

```python
#tuple() Example
x = [1, 2, 3, 4]

y = tuple(x)

print(y)
print(type(y))
```

```python
#set() Example
x = [1, 2, 2, 3, 4, 4]

y = set(x)

print(y)
print(type(y))
```

```python
#dict() Example
x = [('name', 'Vrusha'), ('age', 22)]

y = dict(x)

print(y)
print(type(y))
```

```python
#str to int
age = "22"
age = int(age)
print(age)
print(type(age))
```

```python
#Task 5: Convert "100" to an Integer
num = "100"

print("Before Conversion:")
print(num)
print(type(num))

num = int(num)

print("\nAfter Conversion:")
print(num)
print(type(num))
```
