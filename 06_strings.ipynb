Definition of Strings (Python):

A string is a sequence of characters enclosed within single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """) used to store and manipulate text in Python.

A string can contain letters, numbers, symbols, and spaces.
Strings are immutable, which means their contents cannot be changed after they are created.
Each character in a string has an index, starting from 0.

Indexing
Indexing is used to access individual characters of a string using their position.

```python
#Task 1: Print first, third, last and second-last character.
name = "Python"

print("First Character :", name[0])
print("Third Character :", name[2])
print("Last Character :", name[-1])
print("Second Last Character :", name[-2])
```

```python
"""Task 2: Take a Name as Input and Print
First letter
Middle letter
Last letter"""

name = input("Enter your name: ")

length = len(name)
middle = length // 2

print("First Letter :", name[0])
print("Middle Letter:", name[middle])
print("Last Letter  :", name[-1])
```

```python
#Task 3: Print Every Character of "Data Science" Using Indexing
text = "Data Science"

print(text[0])
print(text[1])
print(text[2])
print(text[3])
print(text[4])
print(text[5])
print(text[6])
print(text[7])
print(text[8])
print(text[9])
print(text[10])
print(text[11])
```

Slicing
Slicing is used to extract a portion of a string using start and end indices.

```python
"""Task 1: From "Artificial Intelligence" print:
"Artificial"
"Intelligence" """

text = "Artificial Intelligence"

print(text[:10])
print(text[11:])
```

```python
#Task 2: Print the First 5 Characters
text = "Artificial Intelligence"

print(text[:5])
```

```python
#Task 3: Print the Last 5 Characters
text = "Artificial Intelligence"

print(text[-5:])
```

```python
#Task 4: Print Every Alternate Character
text = "Artificial Intelligence"

print(text[::2])
```

```python
#Task 5: Reverse the String Using Slicing
text = "Artificial Intelligence"

print(text[::-1])
```

upper() and lower()
upper() converts all characters to uppercase.
lower() converts all characters to lowercase.

```python
#Task 1: Convert "Python Programming" to Uppercase
text = "Python Programming"

print(text.upper())
```

```python
#Task 2: Convert "DATA SCIENCE" to Lowercase
text = "DATA SCIENCE"

print(text.lower())
```

```python
#Task 3: Take User Input and Print Both Uppercase and Lowercase Versions
text = input("Enter any text: ")

print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
```

replace()
replace() replaces one substring with another.

```python
#Task 1: Replace "Python" with "Java" in "Python is easy"
text = "Python is easy"

new_text = text.replace("Python", "Java")

print(new_text)
```

```python
#Task 2: Replace Spaces with Hyphens
text = "Data Science is Awesome"

new_text = text.replace(" ", "-")

print(new_text)
```

```python
#Task 3: Replace Every "a" with "@"
text = "Data Science"

new_text = text.replace("a", "@")

print(new_text)
```

find()
find() returns the index of the first occurrence of a substring, or -1 if it is not found.

```python
#Task 1: Find "Data" in "Data Science"
text = "Data Science"

position = text.find("Data")

print("Position:", position)
```

```python
#Task 2: Find "Python" in "I love Python Programming"
text = "I love Python Programming"

position = text.find("Python")

print("Position:", position)
```

```python
#Task 3: Search for a User-Entered Character in a String
text = input("Enter a string: ")
char = input("Enter a character to search: ")

position = text.find(char)

print("Position:", position)
```

count()
count() returns the number of times a substring appears in a string.

```python
#Task 1: Count the Number of "a" Characters in "banana"
text = "banana"

count = text.count("a")

print("Number of 'a' characters:", count)
```

```python
#Task 2: Count the Spaces in a Sentence
sentence = input("Enter a sentence: ")

spaces = sentence.count(" ")

print("Number of spaces:", spaces)
```

```python
#Task 3: Count How Many Times "Python" Appears in a Sentence
sentence = input("Enter a sentence: ")

count = sentence.count("Python")

print("Number of times 'Python' appears:", count)
```

startswith() and endswith()
startswith() checks whether a string begins with a specified value.
endswith() checks whether a string ends with a specified value.

```python
#Task 1: Check Whether "Python Programming" Starts with "Python"
text = "Python Programming"

result = text.startswith("Python")

print(result)
```

```python
#Task 2: Check Whether "resume.pdf" Ends with ".pdf"
filename = "resume.pdf"

result = filename.endswith(".pdf")

print(result)
```

```python
#Task 3: Take a Filename as Input and Check Whether It Ends with ".txt"
filename = input("Enter a filename: ")

print(filename.endswith(".txt"))
```

f-Strings
f-strings are used to insert variables or expressions directly into strings using {}.

```python
#Task 1: Take Name and Age as Input
name = input("Enter your name: ")
age = int(input("Enter your age: "))

print(f"My name is {name} and I am {age} years old.")
```

```python
#Task 2: Take Two Numbers and Print
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

sum = num1 + num2

print(f"The sum of {num1} and {num2} is {sum}.")
```

```python
#Task 3: Display a Student's Name, Course, and CGPA Using an f-string
name = input("Enter student name: ")
course = input("Enter course: ")
cgpa = float(input("Enter CGPA: "))

print(f"Student Name: {name}")
print(f"Course: {course}")
print(f"CGPA: {cgpa}")
```
