Nested if
Definition: An if statement placed inside another if statement.

```python
"""Task 1: Check Admission Eligibility

Conditions:

Marks ≥ 75
Attendance ≥ 80%"""

marks = float(input("Enter your marks: "))
attendance = float(input("Enter attendance percentage: "))

if marks >= 75:
    if attendance >= 80:
        print("Admission Eligible")
    else:
        print("Not Eligible (Attendance is less than 80%)")
else:
    print("Not Eligible (Marks are less than 75)")
```

```python
"""Task 2: Check Login

Conditions:

Username = "admin"
Password = "1234" """
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Incorrect Password")
else:
    print("Invalid Username")
```

```python
"""Task 3: Check Scholarship Eligibility

Conditions:

Marks ≥ 85
Annual Income ≤ ₹3,00,000"""
marks = float(input("Enter marks: "))
income = int(input("Enter annual income: "))

if marks >= 85:
    if income <= 300000:
        print("Scholarship Eligible")
    else:
        print("Not Eligible (Income too high)")
else:
    print("Not Eligible (Marks too low)")
```

```python
"""Task 4: Check Bank Loan Eligibility

Conditions:

Salary ≥ ₹30,000
Credit Score ≥ 700"""
salary = int(input("Enter monthly salary: "))
credit_score = int(input("Enter credit score: "))

if salary >= 30000:
    if credit_score >= 700:
        print("Loan Approved")
    else:
        print("Loan Rejected (Low Credit Score)")
else:
    print("Loan Rejected (Salary too low)")
```

```python
"""Task 8: Check Employee Promotion

Conditions:

Experience ≥ 5 years
Performance Rating ≥ 4"""
experience = int(input("Enter years of experience: "))
rating = float(input("Enter performance rating (1-5): "))

if experience >= 5:
    if rating >= 4:
        print("Promotion Approved")
    else:
        print("Promotion Denied (Low Performance Rating)")
else:
    print("Promotion Denied (Insufficient Experience)")
```
