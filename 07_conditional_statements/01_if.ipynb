if
Definition: Executes a block of code only if the given condition is True.

```python
#Task 1: Check Whether a Number is Positive
num = int(input("Enter a number: "))

if num > 0:
    print("The number is positive.")
```

```python
#Task 2: Check Whether a Person is Eligible to Vote
age = int(input("Enter your age: "))

if age >= 18:
    print("Eligible to vote.")
```

```python
#Task 3: Check Whether a Character is Uppercase
ch = input("Enter a character: ")

if ch.isupper():
    print("The character is uppercase.")
```

```python
#Task 4: Check Whether a Password Length is at Least 8 Characters
password = input("Enter your password: ")

if len(password) >= 8:
    print("Valid password length.")
```

```python
#Task 5: Check Whether a Number is a Multiple of 10
num = int(input("Enter a number: "))

if num % 10 == 0:
    print("The number is a multiple of 10.")
```
