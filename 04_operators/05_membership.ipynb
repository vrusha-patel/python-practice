Membership Operators 
Definition: Membership operators check whether a value exists in a sequence like a string, list, tuple, set, or dictionary.

Operators:

in → Returns True if the value exists in the collection. (Example: "Python" in ["Python", "Java"] → True)
not in → Returns True if the value does not exist in the collection. (Example: "C++" not in ["Python", "Java"] → True)

```python
"""Task 1

Create a list of fruits.
Ask the user to enter a fruit.
Check whether it exists."""

fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
fruit = input("Enter a fruit: ")

print(fruit in fruits)
```

```python
"""Task 2: 
Check whether a letter exists inside a word"""

word = "Python"
letter = input("Enter a letter: ")
print(letter in word)
```

```python
"""Task 3: 
Check whether a subject exists in a tuple"""

subjects = ("Maths", "Science", "English", "Computer")
subject = input("Enter a subject: ")
print(subject in subjects)
```

```python
"""Task 4: 
Check whether a city is NOT present in a list"""

cities = ["Ahmedabad", "Surat", "Vadodara", "Rajkot"]
city = input("Enter a city: ")
print(city not in cities)
```
