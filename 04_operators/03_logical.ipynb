Logical Operators 

Definition: Logical operators combine or modify multiple conditions and return True or False.

Operators:
and → Returns True only if both conditions are True. (Example: 5 > 2 and 8 > 4 → True)
or → Returns True if at least one condition is True. (Example: 5 > 10 or 8 > 4 → True)
not → Reverses the Boolean result. (Example: not(True) → False)

```python
"""Task 1

Check login.
Input
Username
Password
Compare using logical operators."""

correct_username = "admin"
correct_password = "1234"

username = input("Enter Username: ")
password = input("Enter Password: ")

print(username == correct_username and password == correct_password)
```

```python
"""Task 2

Check admission eligibility.
Condition
Marks ≥75
AND
Attendance ≥80"""

marks = int(input("Enter Marks: "))
attendance = int(input("Enter Attendance (%): "))

print(marks >= 75 and attendance >= 80)
```

```python
"""Task 3

Check holiday.
Saturday OR Sunday"""

day = input("Enter the day: ")

print(day == "Saturday" or day == "Sunday")
```

```python
"""Not Operator"""


is_logged_in = False
print(not is_logged_in)

is_admin = True
print(not is_admin)
```
