Identity Operators
Definition: Identity operators check whether two variables refer to the same object in memory.

Operators:

is → Returns True if both variables point to the same object. (Example: a is b)
is not → Returns True if both variables point to different objects. (Example: a is not b)

Remember: == compares values, while is compares memory locations (object identity).

```python
#Task 1: Compare two lists using == and is

list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 == list2)
print(list1 is list2)
```

```python
#Task 2: Create a = [1,2,3] and b = a

a = [1, 2, 3]
b = a

print(a is b)
print(a == b)
```

```python
#Task 3: Compare two variables pointing to the same string

str1 = "Python"
str2 = str1

print(str1 is str2)
print(str1 == str2)
```

```python
#Task 4: Use is not with two different variables

x = [10, 20]
y = [10, 20]

print(x is not y)
```
