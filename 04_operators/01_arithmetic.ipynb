Arithmetic Operators 
Definition: Arithmetic operators are used to perform mathematical calculations on numbers.

Operators:

+ → Adds two numbers. (Example: 10 + 5 → 15)
- → Subtracts one number from another. (Example: 10 - 5 → 5)
* → Multiplies two numbers. (Example: 10 * 5 → 50)
/ → Divides one number by another and returns a float. (Example: 10 / 5 → 2.0)
// → Divides and returns only the whole-number quotient. (Example: 10 // 3 → 3)
% → Returns the remainder after division. (Example: 10 % 3 → 1)
** → Raises a number to the power of another. (Example: 2 ** 3 → 8)

```python
"""Task 1

Take two integers and print:

Addition
Subtraction
Multiplication
Division"""

a=int(input("Enter 1st number:"))
b=int(input("Enter 2nd number:"))
print("a =",a,"and b =",b)
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
"""Task 2

Print:

Floor Division
Modulus
Exponent"""

print(a//b)
print(a%b)
print(b**a)
```

```python
"""Task 3

Find the area and perimeter of a rectangle.

Input:
Length
Width"""

length=int(input("Enter length:"))
breadth=int(input("Enter breadth:"))

area=length*breadth

print("Area of Rectangle is ",area)
```

```python
"""Task 4

Find Simple Interest."""

Principle_amt = 100000
Rate = 2
Time = 1

S_Int =(Principle_amt*Rate*Time)/100

print("The Simple Interest is ",S_Int)
```

```python
"""Convert total days into:

Years
Months
Remaining Days"""

Total_Days = int(input('Enter no. of days:'))

Years = Total_Days//365
rem_days = Total_Days%365

Months = rem_days//30
Days = rem_days%30

print("Years:",Years)
print("Months:",Months)
print("Days:",Days)
```
