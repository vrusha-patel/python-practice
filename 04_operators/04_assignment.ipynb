Assignment operators
Definition: Assignment operators assign or update values stored in variables.

Operators:
= → Assigns a value to a variable. (Example: x = 10)
+= → Adds and assigns the result. (Example: x += 5)
-= → Subtracts and assigns the result. (Example: x -= 2)
*= → Multiplies and assigns the result. (Example: x *= 3)
/= → Divides and assigns the result. (Example: x /= 2)
//= → Performs floor division and assigns the result. (Example: x //= 2)
%= → Finds the remainder and assigns the result. (Example: x %= 3)
**= → Raises to a power and assigns the result. (Example: x **= 2)

```python
"""Task 1
Take a number.
Perform every assignment operator one by one."""

x=int(input("Enter number"))
x+=5
print(x)
x-=5
print(x)
x*=2
print(x)
x/=2
print(x)
x//=8   
print(x)
x%=3
print(x)
x**=5
print(x)
```

```python
"""Task 2

Calculate monthly savings.
Initial savings = 0
Each month add savings using +="""

savings = 0

month1 = int(input("Enter savings for Month 1: "))
savings += month1

month2 = int(input("Enter savings for Month 2: "))
savings += month2

month3 = int(input("Enter savings for Month 3: "))
savings += month3

print("Total Savings:", savings)
```

```python
"""Task 3
Reduce stock quantity using -="""

stock = 100
sold = int(input("Enter sold quantity: "))
stock -= sold

print("Remaining Stock:", stock)
```
