Bitwise Operators
Definition: Bitwise operators perform operations on the binary (bit) representation of integers.

Operators:

& → Performs bitwise AND. (Example: 5 & 3 → 1)
| → Performs bitwise OR. (Example: 5 | 3 → 7)
^ → Performs bitwise XOR. (Example: 5 ^ 3 → 6)
~ → Performs bitwise NOT. (Example: ~5 → -6)
<< → Shifts bits to the left. (Example: 5 << 1 → 10)
>> → Shifts bits to the right. (Example: 8 >> 1 → 4)

```python
#Task 1: Perform &, |, ^

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("AND (&):", a & b)
print("OR (|):", a | b)
print("XOR (^):", a ^ b)
```

```python
#Task 2: Left Shift (<<)

num = 5

print("5 << 1 =", num << 1)
print("5 << 2 =", num << 2)
```

```python
#Task 3: Right Shift (>>)

num = 20

print("20 >> 1 =", num >> 1)
print("20 >> 2 =", num >> 2)
```

```python
#Task 4: Bitwise NOT (~)

num = 5
print("Bitwise NOT of 5 =", ~num)
```

```python
#Task 5: Observe Binary Representation using bin()

num = int(input("Enter a number: "))
print("Binary representation:", bin(num))
```
