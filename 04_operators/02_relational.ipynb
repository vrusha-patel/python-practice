Relational Operators
Definition: Comparison operators compare two values and always return True or False.

Operators:
== → Checks if two values are equal. (Example: 10 == 10 → True)
!= → Checks if two values are not equal. (Example: 10 != 5 → True)
> → Checks if the left value is greater than the right value. (Example: 10 > 5 → True)
< → Checks if the left value is smaller than the right value. (Example: 5 < 10 → True)
>= → Checks if the left value is greater than or equal to the right value. (Example: 10 >= 10 → True)
<= → Checks if the left value is less than or equal to the right value. (Example: 5 <= 10 → True)

```python
"""Task 1

Take two numbers.
Print every comparison result."""

a=int(input("Number 1:"))
b=int(input("Number 2:"))

print("==",a==b)
print("!=",a!=b)
print(">",a>b)
print("<",a<b)
print(">=",a>=b)
print("<=",a<=b)
```
